# 🧠 Single Agent Pipeline Project

## Problem Statement
Build a **Single-Agent Smart Assistant** that:
- Understands user queries
- Routes tasks based on intent
- Uses tools when required
- Returns structured JSON output

### The agent should handle:
- Math queries → Calculator Tool
- Keyword extraction → Keyword Tool
- General queries → Direct response

---
### 🛠️ What You Need to Implement
- Agent logic
- Conditional routing
- Tool integration
- Basic error handling

### 🚀 Bonus
- Improve routing
- Add logging
- Add more tools


````python
# 🛠️ TOOL 1: Calculator

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        return str(eval(expression))
    except Exception:
        return "Error in calculation"
````

````python
# 🛠️ TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        words = text.split()
        keywords = list(set([w.lower() for w in words if len(w) > 4]))
        return keywords[:5]
    except Exception:
        return []
````

## 🤖 Implement Agent Logic Below

👉 Use conditional routing:
- If query contains "calculate" → use calculator
- If query contains "keywords" → use keyword extractor
- Else → general response

````python
# 🤖 AGENT FUNCTION (TO IMPLEMENT)
import re
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)
logger = logging.getLogger(__name__)

def extract_expression(query: str) -> str:
    match = re.search(r"[\d\s\+\-\*\/\.\(\)]+", query)
    return match.group().strip() if match else query

def agent(query: str):
    query_lower = query.lower()
    logger.info(f"Received query: {query}")

    try:
        if any(word in query_lower for word in
         ["calculate", "solve", "add", "subtract", "multiply", "divide"]):
            logger.info("Routing to: Calculator Tool")
            expression = extract_expression(query)
            result = calculator(expression)
            return {
                "type": "calculation",
                "query": query,
                "expression": expression,
                "result": result
            }

        elif any(word in query_lower for word in
         ["keyword", "keywords", "extract"]):
            logger.info("Routing to: Keyword Extractor Tool")
            result = extract_keywords(query)
            return {
                "type": "keywords",
                "query": query,
                "result": result
            }

        else:
            logger.info("Routing to: General Response")
            return {
                "type": "general",
                "query": query,
                "result": f"You asked: '{query}'. This is a general query handled directly by the agent."
            }

    except Exception as e:
        logger.error(f"Error processing query: {e}")
        return {
            "type": "error",
            "query": query,
            "result": str(e)
        }
````

## 📦 Expected Output Format

```
{
  "type": "calculation / keywords / general / error",
  "result": ...
}
```
