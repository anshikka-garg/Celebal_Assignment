```python
df_clean = df.copy()

print("Original shape:", df.shape)
print("Clean copy shape:", df_clean.shape)
```

```python
numeric_cols = [
    'Year',
    'Month', 
    'Estimated_Deliveries',
    'Production_Units',
    'Avg_Price_USD',
    'Battery_Capacity_kWh',
    'Range_km',
    'CO2_Saved_tons',
    'Charging_Stations'
]

for col in numeric_cols:
    df_clean[col] = pd.to_numeric(df_clean[col], errors='coerce')

print("Data types after fixing:")
print(df_clean.dtypes)
```

```python

## find the misssing values 
missing_counts = df_clean.isnull().sum()

total_missing = df_clean.isnull().sum().sum()

print("Missing values per column:")
print(missing_counts)
print(f"\nTotal missing values in dataset: {total_missing}")

print("\nOnly columns with missing values:")
print(missing_counts[missing_counts > 0])
```

```python
# handling the missing values

numerical_cols_to_fill = [
    'Estimated_Deliveries',
    'Production_Units', 
    'Avg_Price_USD',
    'Battery_Capacity_kWh',
    'Range_km',
    'CO2_Saved_tons',
    'Charging_Stations'
]

for col in numerical_cols_to_fill:
    median_val = df_clean[col].median()
  
    df_clean[col].fillna(median_val, inplace=True)
    
    print(f"Filled '{col}' missing values with median: {median_val:.2f}")

for col in ['Region', 'Model', 'Source_Type']:
    mode_val = df_clean[col].mode()[0]
    df_clean[col].fillna(mode_val, inplace=True)
    print(f"Filled '{col}' missing values with mode: {mode_val}")
```

```python
## confirming no missing values remain


print("\nMissing values after cleaning:")
print(df_clean.isnull().sum())
print(f"\nTotal missing: {df_clean.isnull().sum().sum()}")
```

```python
# duplicate value (find & remove)

num_duplicates = df_clean.duplicated().sum()
print(f"Number of duplicate rows: {num_duplicates}")

df_clean.drop_duplicates(keep='first', inplace=True)

print(f"Shape after removing duplicates: {df_clean.shape}")
```

```python
## checking outliers 

Q1 = df_clean['Estimated_Deliveries'].quantile(0.25)

Q3 = df_clean['Estimated_Deliveries'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Q1 (25th percentile): {Q1:,.0f}")
print(f"Q3 (75th percentile): {Q3:,.0f}")
print(f"IQR: {IQR:,.0f}")
print(f"Lower bound: {lower_bound:,.0f}")
print(f"Upper bound: {upper_bound:,.0f}")

outliers = df_clean[
    (df_clean['Estimated_Deliveries'] < lower_bound) | 
    (df_clean['Estimated_Deliveries'] > upper_bound)
]

print(f"\nNumber of outlier rows: {len(outliers)}")
print("\nSample outlier rows:")
print(outliers[['Year', 'Month', 'Model', 'Region', 
                'Estimated_Deliveries']].head(10))
```

```python
df_clean['Estimated_Deliveries'] = np.clip(
    df_clean['Estimated_Deliveries'],  
    lower_bound,                        
    upper_bound                         
)

print("Deliveries after outlier capping:")
print(df_clean['Estimated_Deliveries'].describe())
```

```python
reliability_map = {
    'Official (Quarter)' : 3,   
    'Interpolated (Month)': 2,  
    'Estimated (Region)' : 1    
}

df_clean['Data_Reliability'] = df_clean['Source_Type'].map(reliability_map)

print("Data reliability scores added:")
print(df_clean['Data_Reliability'].value_counts())
```

```python
df_feat = pd.read_csv('tesla_prepared.csv')

df_feat['Quarter'] = np.ceil(df_feat['Month'] / 3).astype(int)

df_feat['is_quarter_end'] = df_feat['Month'].isin([3, 6, 9, 12]).astype(int)

df_feat['years_active'] = df_feat['Year'] - 2003

print("New date features:")
print(df_feat[['Year', 'Month', 'Quarter',
               'is_quarter_end', 'years_active']].head(12))
```

```python
df_feat = df_feat.sort_values(['Year', 'Month', 'Model_Model S']).reset_index(drop=True)

df_feat['lag_1_deliveries'] = df_feat['Estimated_Deliveries'].shift(1)

df_feat['lag_3_deliveries'] = df_feat['Estimated_Deliveries'].shift(3)

df_feat['lag_12_deliveries'] = df_feat['Estimated_Deliveries'].shift(12)

print("Lag features created:")
print(df_feat[['Estimated_Deliveries',
               'lag_1_deliveries',
               'lag_3_deliveries',
               'lag_12_deliveries']].head(15))
```

```python
## rolling statistics or rolling window 
df_feat['rolling_mean_3'] = (
    df_feat['Estimated_Deliveries']
    .rolling(window=3, min_periods=1)
    .mean()
)

df_feat['rolling_mean_6'] = (
    df_feat['Estimated_Deliveries']
    .rolling(window=6, min_periods=1)
    .mean()
)

df_feat['rolling_std_3'] = (
    df_feat['Estimated_Deliveries']
    .rolling(window=3, min_periods=1)
    .std()
)

print("Rolling features:")
print(df_feat[['Estimated_Deliveries',
               'rolling_mean_3',
               'rolling_mean_6',
               'rolling_std_3']].head(10))
```

```python
df_feat['production_efficiency'] = (
    df_feat['Estimated_Deliveries'] /
    (df_feat['Production_Units'] + 1)
)

df_feat['price_per_km'] = (
    df_feat['Avg_Price_USD'] /
    (df_feat['Range_km'] + 1)
)

print("Interaction features:")
print(df_feat[['Estimated_Deliveries',
               'production_efficiency',
               'price_per_km']].head(8))
```

```python
print(f"Shape before dropping NaN: {df_feat.shape}")

df_feat.dropna(inplace=True)

df_feat.reset_index(drop=True, inplace=True)

print(f"Shape after dropping NaN:  {df_feat.shape}")
print(f"Rows removed: {2640 - len(df_feat)}")
```

```python
# all features 
cols_to_drop = ['Estimated_Deliveries', 'Source_Type']

cols_to_drop = [c for c in cols_to_drop if c in df_feat.columns]

X = df_feat.drop(columns=cols_to_drop)

y = df_feat['Estimated_Deliveries']

print(f"Features (X) shape: {X.shape}")
print(f"Target (y) shape:   {y.shape}")
print(f"\nAll feature names:")
for i, col in enumerate(X.columns):
    print(f"  {i+1:2d}. {col}")

df_feat.to_csv('tesla_features.csv', index=False)
print("\nSaved as tesla_features.csv")
```

```python
##removing the leaky features
df_model = pd.read_csv('tesla_features.csv')

leaky_cols = ['CO2_Saved_tons', 'production_efficiency']

leaky_cols = [c for c in leaky_cols if c in df_model.columns]
df_model.drop(columns=leaky_cols, inplace=True)

print("Leaky columns removed:", leaky_cols)
print(f"Shape after leakage removal: {df_model.shape}")
```

```python
#Chronological train-test split 

df_model = df_model.sort_values(['Year', 'Month']).reset_index(drop=True)

split_year = 2023

train_mask = df_model['Year'] < split_year
test_mask  = df_model['Year'] >= split_year

train_df = df_model[train_mask].reset_index(drop=True)
test_df  = df_model[test_mask].reset_index(drop=True)

print(f"Training set: {train_df['Year'].min()} to {train_df['Year'].max()}")
print(f"Training rows: {len(train_df)}")
print(f"\nTest set: {test_df['Year'].min()} to {test_df['Year'].max()}")
print(f"Test rows: {len(test_df)}")
print(f"\nTrain/Test split: {len(train_df)/len(df_model)*100:.1f}% / "
      f"{len(test_df)/len(df_model)*100:.1f}%")
```

```python
# ML Pipeline 

from sklearn.pipeline import Pipeline

from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LinearRegression

target = 'Estimated_Deliveries'

drop_cols = [target, 'Source_Type'] if 'Source_Type' in df_model.columns \
             else [target]

feature_cols = [c for c in df_model.columns if c not in drop_cols]

X_train = train_df[feature_cols]
y_train = train_df[target]

X_test  = test_df[feature_cols]
y_test  = test_df[target]

print(f"X_train shape: {X_train.shape}")
print(f"X_test shape:  {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape:  {y_test.shape}")

pipeline_lr = Pipeline([
    ('scaler', StandardScaler()),

    ('model', LinearRegression())
])

pipeline_lr.fit(X_train, y_train)

print("Pipeline trained successfully!")
print("\nPipeline steps:")
for name, step in pipeline_lr.steps:
    print(f"  {name}: {step}")
```

```python
# comparing actual vs predicted and finding the difference 
y_pred_lr = pipeline_lr.predict(X_test)

print(f"{'Actual':>12} {'Predicted':>12} {'Difference':>12}")
print("-" * 38)
for actual, predicted in zip(y_test[:10], y_pred_lr[:10]):
    diff = actual - predicted
    print(f"{actual:>12,.0f} {predicted:>12,.0f} {diff:>12,.0f}")
```

```python
# Ridge Regression 

from sklearn.linear_model import Ridge

pipeline_ridge = Pipeline([
    ('scaler', StandardScaler()),
    ('model', Ridge(alpha=1.0))
])

pipeline_ridge.fit(X_train, y_train)
y_pred_ridge = pipeline_ridge.predict(X_test)
```

```python
#Lasso Regression pipeline

from sklearn.linear_model import Lasso

pipeline_lasso = Pipeline([
    ('scaler', StandardScaler()),
    ('model', Lasso(alpha=1.0, max_iter=10000))
])

pipeline_lasso.fit(X_train, y_train)
y_pred_lasso = pipeline_lasso.predict(X_test)

lasso_coefs = pipeline_lasso.named_steps['model'].coef_

print("\nLasso feature weights:")
for feature, coef in zip(feature_cols, lasso_coefs):
    status = "KEPT" if abs(coef) > 0.001 else "REMOVED"
    print(f"  {status:8s} {feature:35s}: {coef:.4f}")
```

```python
#Evaluate all  models 

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

def evaluate_model(name, y_true, y_pred):

    mae = mean_absolute_error(y_true, y_pred)

    mse = mean_squared_error(y_true, y_pred)

    rmse = np.sqrt(mse)

    r2 = r2_score(y_true, y_pred)

    mape = np.mean(np.abs((y_true - y_pred) / (y_true + 1))) * 100

  
    print(f"Model: {name}")
   
    print(f"  MAE  : {mae:>10,.2f}  (avg error in deliveries)")
    print(f"  RMSE : {rmse:>10,.2f}  (penalises large errors more)")
    print(f"  R²   : {r2:>10.4f}  (1.0 = perfect)")
    print(f"  MAPE : {mape:>10.2f}% (avg % error)")

    return {'name': name, 'MAE': mae, 'RMSE': rmse, 'R2': r2, 'MAPE': mape}


results = []
results.append(evaluate_model("Linear Regression", y_test, y_pred_lr))
results.append(evaluate_model("Ridge Regression",  y_test, y_pred_ridge))
results.append(evaluate_model("Lasso Regression",  y_test, y_pred_lasso))


print("\n\nMODEL  SUMMARY")
print(f"{'Model':<25} {'MAE':>10} {'RMSE':>10} {'R²':>8} {'MAPE':>8}")

for r in results:
    print(f"{r['name']:<25} {r['MAE']:>10,.0f} "
          f"{r['RMSE']:>10,.0f} {r['R2']:>8.4f} {r['MAPE']:>7.1f}%")
```

```python
#Time Series Cross Validation

from sklearn.model_selection import TimeSeriesSplit, cross_val_score

tscv = TimeSeriesSplit(n_splits=5)
cv_scores_lr = cross_val_score(
    pipeline_lr,          
    X_train,              
    y_train,              
    cv=tscv,              
    scoring='neg_root_mean_squared_error'
)

cv_rmse_lr = -cv_scores_lr

print("Linear Regression — Cross Validation RMSE per fold:")
for i, score in enumerate(cv_rmse_lr):
    print(f"  Fold {i+1}: {score:,.2f}")
print(f"  Mean RMSE: {cv_rmse_lr.mean():,.2f}")
print(f"  Std RMSE:  {cv_rmse_lr.std():,.2f}")

cv_rmse_ridge = -cross_val_score(
    pipeline_ridge, X_train, y_train,
    cv=tscv,
    scoring='neg_root_mean_squared_error'
)

cv_rmse_lasso = -cross_val_score(
    pipeline_lasso, X_train, y_train,
    cv=tscv,
    scoring='neg_root_mean_squared_error'
)

print(f"\nRidge CV  — Mean RMSE: {cv_rmse_ridge.mean():,.2f} "
      f"± {cv_rmse_ridge.std():,.2f}")
print(f"Lasso CV  — Mean RMSE: {cv_rmse_lasso.mean():,.2f} "
      f"± {cv_rmse_lasso.std():,.2f}")
```

```python
# Hyperparameter Tuning 
#ridge

from sklearn.model_selection import GridSearchCV
param_grid = {
    'model__alpha': [0.01, 0.1, 1.0, 10.0, 100.0, 1000.0]
}

grid_search_ridge = GridSearchCV(
    pipeline_ridge,      
    param_grid,           
    cv=tscv,              
    scoring='neg_root_mean_squared_error',
    verbose=1,         
    n_jobs=-1            
)

grid_search_ridge.fit(X_train, y_train)

print(f"\nBest alpha: {grid_search_ridge.best_params_}")
print(f"Best CV RMSE: {-grid_search_ridge.best_score_:,.2f}")

##lasso
param_grid_lasso = {
    'model__alpha': [0.01, 0.1, 1.0, 10.0, 100.0]
}

grid_search_lasso = GridSearchCV(
    pipeline_lasso,
    param_grid_lasso,
    cv=tscv,
    scoring='neg_root_mean_squared_error',
    verbose=1,
    n_jobs=-1
)

grid_search_lasso.fit(X_train, y_train)

print(f"\nBest alpha for Lasso: {grid_search_lasso.best_params_}")
print(f"Best CV RMSE: {-grid_search_lasso.best_score_:,.2f}")
```

```python
# ADF Test

from statsmodels.tsa.stattools import adfuller
monthly_ts = (
    df_model.groupby(['Year', 'Month'])['Estimated_Deliveries']
    .sum()
    .reset_index()
    .sort_values(['Year', 'Month'])
)

adf_result = adfuller(monthly_ts['Estimated_Deliveries'])

print("ADF Test for Stationarity:")
print(f"  Test Statistic : {adf_result[0]:.4f}")
print(f"  p-value        : {adf_result[1]:.4f}")
print(f"  Critical Values:")
for key, val in adf_result[4].items():
    print(f"    {key}: {val:.4f}")

print()
if adf_result[1] < 0.05:
    print("p-value < 0.05 → STATIONARY ")
    print("Mean and variance are stable over time")
else:
    print("p-value > 0.05 → NOT STATIONARY ")
    print("Series has a trend — we handle this via lag features")
    print("Our lag features already account for this!")
```
